```python
# 1. Import libraries
import tensorflow as tf
import torch

# 2. TensorFlow operations
a = tf.constant([10, 20, 30])
b = tf.constant([5, 10, 15])

print("TensorFlow Operations")
print("Addition:", tf.add(a, b).numpy())
print("Subtraction:", tf.subtract(a, b).numpy())
print("Multiplication:", tf.multiply(a, b).numpy())

# 3. TensorFlow matrix multiplication
m1 = tf.constant([[1, 2], [3, 4]])
m2 = tf.constant([[5, 6], [7, 8]])

print("Matrix Multiplication:")
print(tf.matmul(m1, m2).numpy())

# 4. PyTorch operations
x = torch.tensor([10, 20, 30])
y = torch.tensor([5, 10, 15])

print("\nPyTorch Operations")
print("Addition:", torch.add(x, y))
print("Subtraction:", torch.sub(x, y))
print("Multiplication:", torch.mul(x, y))

# 5. PyTorch matrix multiplication
p1 = torch.tensor([[1, 2], [3, 4]])
p2 = torch.tensor([[5, 6], [7, 8]])

print("Matrix Multiplication:")
print(torch.matmul(p1, p2))

# 6. Tensor properties
print("\nTensor Properties")
print("Shape:", x.shape)
print("Dimension:", x.ndim)
print("Data Type:", x.dtype)
print("NumPy Array:", x.numpy())
```
