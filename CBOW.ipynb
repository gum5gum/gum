```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Lambda, Dense
import tensorflow as tf
```

```python
data = ["the cat eats fish"]
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(data)
```

```python
words = tokenizer.word_index
vocab_size = len(words) + 1
```

```python
# 3. Create context and target
seq = tokenizer.texts_to_sequences(data)[0]

X = []
Y = []

for i in range(1, len(seq)-1):
    X.append([seq[i-1], seq[i+1]])
    Y.append(seq[i])

X = np.array(X)
Y = np.array(Y)
```

```python
# 4. Create CBOW model
model = Sequential([
    Embedding(vocab_size, 10),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    Dense(vocab_size, activation='softmax')
])
```

```python
# 5. Compile model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
model.fit(X, Y, epochs=100, verbose=0)
```

```python
# 7. Predict target word
test = np.array([[words['cat'], words['fish']]])

prediction = model.predict(test, verbose=0)

index = np.argmax(prediction)
predicted_word = tokenizer.index_word[index]

print("Predicted word:", predicted_word)
```
