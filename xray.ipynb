```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
import numpy as np
import matplotlib.pyplot as plt
```

```python
train = r"D:/Jupyter/DL/5/Pneumonia X Ray/train"
test = r"D:/Jupyter/DL/5/Pneumonia X Ray/test"
```

```python
train_gen = ImageDataGenerator(rescale=1./255).flow_from_directory(
    train,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical'
)

test_gen = ImageDataGenerator(rescale=1./255).flow_from_directory(
    test,
    target_size=(224, 224),
    batch_size=32,
    class_mode='categorical'
)
```

```python
class_names = list(train_gen.class_indices.keys())

print(class_names)
```

```python
model = Sequential([
    Conv2D(64, (3, 3), activation='relu', input_shape=(224, 224, 3)),
    MaxPooling2D((2, 2)),

    Conv2D(32, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Conv2D(16, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(128, activation='relu'),
    Dense(2, activation='softmax')
])
```

```python
model.compile(optimizer='adam',loss='categorical_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(train_gen,validation_data=test_gen,epochs=5,batch_size=64,verbose=1)
```

```python
x_test, y_test = next(test_gen)
preds = np.argmax(model.predict(x_test[:5]), axis=1)
actual = np.argmax(y_test[:5], axis=1)
```

```python
print(history.history['accuracy'][-1])
print(history.history['val_accuracy'])
```

```python
plt.figure(figsize=(12, 4))
for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(x_test[i])
    plt.title(f"p:{class_names[preds[i]]}\nA : {class_names[actual[i]]}")
    plt.axis('off')
plt.show()
```

```python
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.legend()
plt.show()
```
